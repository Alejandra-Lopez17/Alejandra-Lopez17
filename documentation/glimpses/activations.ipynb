```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.style.use("ggplot")
plt.figure(figsize=(10, 6))
x = np.linspace(-5, 5, 1000)

step = np.where(x >= 0, 1, 0)
plt.subplot(2, 2, 1)
plt.title("step")
plt.plot(x, step, "gray")
plt.grid(True)
plt.yticks([0, 1])
plt.xticks([-4, 0, 4])

sigmoid = 1 / (1 + np.exp(-x))
plt.subplot(2, 2, 2)
plt.title("sigmoid")
plt.plot(x, sigmoid, "gray")
plt.grid(True)
plt.yticks([0, 0.5, 1])
plt.xticks([-4, 0, 4])

tanh = np.tanh(x)
plt.subplot(2, 2, 3)
plt.title("tanh")
plt.plot(x, tanh, "gray")
plt.grid(True)
plt.yticks([-1, 0, 1])
plt.xticks([-4, 0, 4])

relu = np.maximum(0, x)
plt.subplot(2, 2, 4)
plt.title("ReLU")
plt.plot(x, relu, "gray")
plt.grid(True)
plt.yticks([0, 5])
plt.xticks([0, 5])

plt.tight_layout()
```
