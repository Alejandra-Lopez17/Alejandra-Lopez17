```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
```

```python
x = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 1, 1, 0])

log_reg = LogisticRegression()
log_reg.fit(x, y)

print(f"Accuracy: {log_reg.score(x, y)}")
```

```python
X = np.hstack([x, (x[:, 0] * x[:, 1]).reshape(-1, 1)])
log_reg_poly = LogisticRegression(penalty=None, max_iter=1000)
log_reg_poly.fit(X, y)

print(f"Accuracy with polynomial features: {log_reg_poly.score(X, y)}")
print(f"Coefficients: {log_reg_poly.coef_}")
print(f"Intercept: {log_reg_poly.intercept_}")
```

```python
x_min, x_max = x[:, 0].min() - 0.5, x[:, 0].max() + 0.5
y_min, y_max = x[:, 1].min() - 0.5, x[:, 1].max() + 0.5
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.02), np.arange(y_min, y_max, 0.02))

grid = np.c_[xx.ravel(), yy.ravel()]
grid_poly = np.hstack([grid, (grid[:, 0] * grid[:, 1]).reshape(-1, 1)])

Z = log_reg_poly.predict(grid_poly)
Z = Z.reshape(xx.shape)

plt.style.use("ggplot")
plt.figure(figsize=(10, 6))
plt.contourf(xx, yy, Z, alpha=0.4)
plt.scatter(x[y == 0, 0], x[y == 0, 1], marker="o", s=75, c="black", alpha=0.8)
plt.scatter(x[y == 1, 0], x[y == 1, 1], marker="x", s=75, c="black", alpha=0.8)
```

```python
x_min, x_max = x[:, 0].min() - 0.5, x[:, 0].max() + 0.5
y_min, y_max = x[:, 1].min() - 0.5, x[:, 1].max() + 0.5
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.02), np.arange(y_min, y_max, 0.02))

grid = np.c_[xx.ravel(), yy.ravel()]
grid_poly = np.hstack([grid, (grid[:, 0] * grid[:, 1]).reshape(-1, 1)])

Z = 1 / (1 + np.exp(-(grid_poly @ np.array([14, 14, -28]) + -6)))
Z = (Z > 0.5).reshape(xx.shape)

plt.figure(figsize=(10, 6))
plt.contourf(xx, yy, Z, alpha=0.4)
plt.scatter(x[y == 0, 0], x[y == 0, 1], marker="o", s=75, c="black", alpha=0.8)
plt.scatter(x[y == 1, 0], x[y == 1, 1], marker="x", s=75, c="black", alpha=0.8)
```
