```python
group_a = pd.read_csv("./data/group-a.txt", sep="\t")
group_b = pd.read_csv("./data/group-b.txt", sep="\t")
group_a.head()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
print("Grupo A:")
print(group_a.head())
print("\nGrupo B:")
print(group_b.head())
```

```python
print("\nInformación del Grupo A:")
print(group_a.info())
print("\nInformación del Grupo B:")
print(group_b.info())
```

```python
print("\nEstadísticas descriptivas del Grupo A:")
print(group_a.describe())
print("\nEstadísticas descriptivas del Grupo B:")
print(group_b.describe())
```

```python
print("\nValores nulos en el Grupo A:")
print(group_a.isnull().sum())
print("\nValores nulos en el Grupo B:")
print(group_b.isnull().sum())
```

```python
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
sns.histplot(group_a["bmi"], kde=True, color="blue")
plt.title("Distribución del bmi - Grupo A")
plt.xlabel("bmi")
plt.ylabel("Frecuencia")
```

```python
plt.subplot(1, 2, 2)
sns.histplot(group_b["bmi"], kde=True, color="green")
plt.title("Distribución del bmi - Grupo B")
plt.xlabel("bmi")
plt.ylabel("Frecuencia")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(
    x="bmi", data=pd.concat([group_a.assign(Group="A"), group_b.assign(Group="B")])
)
plt.title("Boxplot del bmi - Grupo A vs Grupo B")
plt.xlabel("bmi")
plt.ylabel("Grupo")
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
sns.scatterplot(x="steps", y="bmi", data=group_a, color="blue")
plt.title("steps vs bmi - Grupo A")
plt.xlabel("Pasos")
plt.ylabel("bmi")
```

```python
plt.subplot(2, 2, 4)
boxplot_data = [group_a["steps"], group_b["steps"]]
plt.boxplot(boxplot_data, tick_labels=["Grupo A", "Grupo B"])
plt.title("Comparación de pasos semanales entre grupos")
plt.ylabel("Pasos")

plt.tight_layout()
plt.show()
```
