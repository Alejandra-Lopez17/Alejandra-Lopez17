```python
df = pd.read_csv("./data/synthetic-polynomial-data.txt", sep="\t")
x = df["x"]
y = df["y"]

plt.style.use("ggplot")
plt.figure(figsize=(10, 6))
plt.scatter(x, y, color="m", marker="x", s=5)
plt.xlabel("x")
plt.ylabel("y")
plt.ylim(top=15000)
plt.xlim(right=55)
plt.show()
```

```python
print(df.head())
print(df.describe())
```

```python
modelo_lineal = SGDRegressor(max_iter=1000, tol=1e-3, random_state=42)
modelo_lineal.fit(x.values.reshape(-1, 1), y)
```

```python
y_pred_lineal = modelo_lineal.predict(x.values.reshape(-1, 1))
```

```python
mse_lineal = mean_squared_error(y, y_pred_lineal)
print(f"Error cuadrático medio (MSE) del modelo lineal: {mse_lineal}")
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(x, y, color="m", marker="x", s=5, label="Datos reales")
plt.plot(x, y_pred_lineal, color="b", label="Predicción lineal")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

El modelo lineal tiene un error muy alto, lo que indica que no es capaz de capturar la relacion no lineal entre x e y

La grafica muestra que la linea de regresion lineal no se ajusta bien a los datos.

Transformacion polinomial

```python
grado_polinomio = 2
poly = PolynomialFeatures(grado_polinomio)
x_poly = poly.fit_transform(x.values.reshape(-1, 1))
```

```python
modelo_polinomico = SGDRegressor(max_iter=1000, tol=1e-3, random_state=42)
modelo_polinomico.fit(x_poly, y)
```

```python
y_pred_poly = modelo_polinomico.predict(x_poly)
```

```python
mse_poly = mean_squared_error(y, y_pred_poly)
print(
    f"Error cuadrático medio (MSE) del modelo polinomial (grado {grado_polinomio}): {mse_poly}"
)
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(x, y, color="m", marker="x", s=5, label="Datos reales")
plt.plot(
    x, y_pred_poly, color="g", label=f"Predicción polinomial (grado {grado_polinomio})"
)
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

Resultados:

    MSE del modelo polinomial (grado 2): 7.46e+29

Analisis:

El error es extremadamente alto, lo que sugiere que el modelo no converge correctamente

Esto puede deberse a la falta de escalado de las características polinomiales, lo que dificulta el entrenamiento del modelo

Experimentar con hiperparametros

```python
tasas_aprendizaje = ["constant", "invscaling", "adaptive"]
for tasa in tasas_aprendizaje:
    modelo = SGDRegressor(max_iter=1000, tol=1e-3, learning_rate=tasa, random_state=42)
    modelo.fit(x_poly, y)
    y_pred = modelo.predict(x_poly)
    mse = mean_squared_error(y, y_pred)
    print(f"MSE con tasa de aprendizaje '{tasa}': {mse}")
```

Resultados:

    MSE con tasa de aprendizaje 'constant': 4.33e+28

    MSE con tasa de aprendizaje 'invscaling': 7.46e+29

    MSE con tasa de aprendizaje 'adaptive': 2.79e+22

Analisis:

La tasa de aprendizaje adaptive tuvo el menor error, pero aún es extremadamente alto

Esto confirma que el modelo no está convergiendo adecuadamente, probablemente debido a la falta de escalado de las características

Comparar diferentes grados polinomiales

```python
grados = [2, 3, 4, 5]
for grado in grados:
    poly = PolynomialFeatures(grado)
    x_poly = poly.fit_transform(x.values.reshape(-1, 1))
    modelo = SGDRegressor(max_iter=1000, tol=1e-3, random_state=42)
    modelo.fit(x_poly, y)
    y_pred = modelo.predict(x_poly)
    mse = mean_squared_error(y, y_pred)
    print(f"MSE con polinomio de grado {grado}: {mse}")
```

Resultados:

    MSE con polinomio de grado 2: 7.46e+29

    MSE con polinomio de grado 3: 3.90e+35

    MSE con polinomio de grado 4: 4.20e+40

    MSE con polinomio de grado 5: 1.14e+50

Analisis:

A medida que aumenta el grado del polinomio, el error aumenta drasticamente.

Esto indica que el modelo esta sobreajustando los datos o que no esta convergiendo debido a la falta de escalado.

proceso de aprendizaje

```python
modelo = SGDRegressor(max_iter=1, tol=None, random_state=42, warm_start=True)
loss_history = []
n_epochs = 1000

for epoch in range(n_epochs):
    modelo.partial_fit(x_poly, y)
    y_pred = modelo.predict(x_poly)  
    loss = mean_squared_error(y, y_pred)  
    loss_history.append(loss)


plt.figure(figsize=(10, 6))
plt.plot(loss_history, label="Pérdida (MSE)")
plt.xlabel("Iteraciones")
plt.ylabel("MSE")
plt.title("Proceso de aprendizaje del SGDRegressor")
plt.legend()
plt.show()
```

Analisis:

La grafica de la perdida muestra que el modelo no converge, ya que el MSE no disminuye significativamente

Esto confirma que el modelo no esta aprendiendo adecuadamente, probablemente debido a la falta de escalado de las caracteristicas

Discusion de tecnicas efectivas para minimizar el error

Escalado de caracteristicas:

    Las caracteristicas polinomiales tienen valores muy grandes, lo que dificulta el entrenamiento del modelo

    Escalar las caracteristicas (por ejemplo, usando StandardScaler) podria ayudar a que el modelo converja más rapidamente

Regularizacion:

    Agregar regularizacion (por ejemplo, L2 o L1) podria evitar el sobreajuste y mejorar la generalizacion del modelo.

Conclusion final

    El modelo lineal no es adecuado para estos datos, ya que no puede capturar la relación no lineal entre x e y.

    La transformacion polinomial mejora la capacidad del modelo para capturar relaciones no lineales, pero el error sigue siendo extremadamente alto debido a la falta de escalado y la posible falta de convergencia del modelo.

    Para mejorar el rendimiento, se recomienda:
