```python
raw_data = load_breast_cancer(as_frame=True)
df = raw_data.data
df["diagnosis"] = raw_data.target
df.head()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Revisa la estructura de los datos
df.info()
df.describe()
```

```python
## valores nulos
df.isnull().sum()
```

```python
# Analiza la distribución de las características numéricas
df.hist(figsize=(20, 15))
plt.show()
```

```python
# Compara la distribución de características entre diagnósticos benignos y maligno
sns.pairplot(
    df, hue="diagnosis", vars=["mean radius", "mean texture", "mean perimeter"]
)
plt.show()
```

```python
# Calcula la matriz de correlación

corr_matrix = df.corr()
plt.figure(figsize=(15, 10))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm")
plt.show()
```

```python
# Crea gráficos de caja para comparar características entre diagnósticos
plt.figure(figsize=(15, 10))
sns.boxplot(data=df, x="diagnosis", y="mean radius")
plt.show()
```

Resumen de Hallazgos
